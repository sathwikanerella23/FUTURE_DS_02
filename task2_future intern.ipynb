```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
data = {

    "Customer_ID": [
        101,102,103,104,105,
        106,107,108,109,110
    ],

    "Gender": [
        "Male","Female","Male","Female","Male",
        "Female","Male","Female","Male","Female"
    ],

    "Subscription_Type": [
        "Basic","Premium","Basic","Standard","Premium",
        "Basic","Standard","Premium","Basic","Standard"
    ],

    "Monthly_Charges": [
        500,1200,450,800,1500,
        400,750,1300,500,850
    ],

    "Tenure_Months": [
        2,24,3,12,36,
        1,10,30,4,15
    ],

    "Churn": [
        "Yes","No","Yes","No","No",
        "Yes","No","No","Yes","No"
    ],

    "Region": [
        "South","North","East","West","South",
        "North","East","West","South","North"
    ]
}
```

```python
df = pd.DataFrame(data)

print(df)
```

```python
total_customers = len(df)

print("="*45)
print("         TOTAL CUSTOMERS")
print("="*45)

print("Total Customers :", total_customers)
```

```python
churn_count = df['Churn'].value_counts()

print("="*45)
print("           CHURN COUNT")
print("="*45)

print(churn_count)
```

```python
subscription_analysis = df.groupby(
    'Subscription_Type'
)['Monthly_Charges'].sum()

print("="*45)
print("      SUBSCRIPTION ANALYSIS")
print("="*45)

print(subscription_analysis)
```

```python
region_analysis = df.groupby(
    'Region'
)['Monthly_Charges'].sum()

print("="*45)
print("         REGION ANALYSIS")
print("="*45)

print(region_analysis)
```

```python
tenure_analysis = df.groupby(
    'Churn'
)['Tenure_Months'].mean()

print("="*45)
print("         TENURE ANALYSIS")
print("="*45)

print(tenure_analysis)
```

```python
plt.figure(figsize=(7,7))

churn_count.plot(
    kind='pie',
    autopct='%1.1f%%'
)

plt.title("Customer Churn Distribution")

plt.ylabel("")

plt.show()
```

```python
plt.figure(figsize=(8,5))

subscription_analysis.plot(kind='bar')

plt.title("Subscription Revenue")

plt.xlabel("Subscription Type")

plt.ylabel("Revenue")

plt.xticks(rotation=0)

plt.show()
```

```python
plt.figure(figsize=(8,5))

region_analysis.plot(kind='bar')

plt.title("Region Performance")

plt.xlabel("Region")

plt.ylabel("Revenue")

plt.xticks(rotation=0)

plt.show()
```

```python
plt.figure(figsize=(6,5))

tenure_analysis.plot(kind='bar')

plt.title("Average Tenure by Churn")

plt.xlabel("Churn")

plt.ylabel("Average Tenure")

plt.xticks(rotation=0)

plt.show()
```


## Insights

- Premium subscription generated highest revenue.
- Customers with low tenure showed higher churn.
- South region experienced more churn cases.
- Long-term customers stayed more active.
# Recommendations

- Improve retention strategies for new customers.
- Offer discounts for long-term subscriptions.
- Increase customer engagement programs.
- Focus on reducing churn in high-risk regions.
